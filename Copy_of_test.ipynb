Simple Load Testing

```python
import requests
import time

url = "https://heart-disease-api-jd4c.onrender.com"

start = time.time()

success = 0

for i in range(100):

    try:
        r = requests.get(url)

        if r.status_code == 200:
            success += 1

    except:
        pass

end = time.time()

print("Total Requests:", 100)
print("Successful:", success)
print("Time Taken:", end-start)
```

Concurrent Users Simulation

```python
import requests
import concurrent.futures
import time

url = "https://heart-disease-api-jd4c.onrender.com"

def hit_api(i):
    try:
        r = requests.get(url)
        return r.status_code
    except:
        return 0

start = time.time()

with concurrent.futures.ThreadPoolExecutor(max_workers=100) as executor:
    results = list(executor.map(hit_api, range(1000)))

end = time.time()

success = results.count(200)

print("Total Requests =", len(results))
print("Successful Requests =", success)
print("Failed Requests =", len(results)-success)
print("Time Taken =", round(end-start,2),"seconds")
```

Bottleneck Analysis

Bottlenecks Identified:

1. Initial cold start delay on free Render instance.
2. Increased latency under heavy concurrent traffic.
3. Single worker deployment limits throughput.

Optimizations Implemented:
